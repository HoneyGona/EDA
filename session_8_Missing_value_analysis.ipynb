**Mean-Median-Mode**
- insted of filling with random values
- we can use mean median and mode also 
- For Numerical columns use mean use Median
    - if Data outliers are there use **Median**
- For categorical columns use **Mode**

```python
age_median=df['Age'].median()
df['Age'].fillna(age_median)
```

```python
age_mode=df['Age'].mode()
df['Age'].fillna(age_mode)
```

**KNN Immputer**
- K- Nearest Neighbours
- Where = hyper parameter which means the user can charge
- Idea: Instead of taking all the samples average
    - First fix the neighbours number i.e.k
    - Then find the neighbours using distance metric
    - Then take the average of those samples to fill the missing value

```python
import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import KNNImputer



```

```python
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
X = [[1,2, np.nan],[3,np.nan,3],[np.nan,60,5] ,[8,8,7]]
pd.DataFrame(X)

```

```python
import numpy as np
from sklearn.impute import KNNImputer
X = [[1,2, np.nan],[3,np.nan,3],[np.nan,60,5] ,[8,8,7]]
imputer = KNNImputer(n_neighbors=2,
                     weights='uniform')
imputer.fit_transform(X)
```

```python
import numpy as np
from sklearn.impute import KNNImputer
X = [[1,2, np.nan],[3,np.nan,3],[np.nan,60,5] ,[8,8,7]]
imputer = KNNImputer(n_neighbors=2,
                     weights='distance')
imputer.fit_transform(X)
```
