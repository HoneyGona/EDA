**Z score**

```python
# step-1 : read pwage column x
# step-2: calculate mean pwage
# step-3: caluculate std of wage
# step-4: pwage-mean
# step-5:visa_df[p_z] step4/step5
```

```python
wage_data=visa_df['prevailing_wage']
wage_mean=wage_data.mean()
wage_std=wage_data.std()
visa_df['prevailing_wage_z']=(wage_data-wage_mean)/wage_std
```
