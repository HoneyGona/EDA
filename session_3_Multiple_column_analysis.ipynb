**bar chart automate**

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.figure(figsize=(10,4))
plt.bar(keys,values)
plt.xlabel('labels')
plt.ylabel('count')
plt.title('continent_barchart')
plt.savefig('continent_bar.jpg')
```

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(10,4))
    plt.bar(keys,values)
    plt.xlabel('labels')
    plt.ylabel('count')
    plt.title(f"{i}.barchart")
    plt.savefig(f"{i}.bar.jpg")
```

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.pie(values,explode=[0.1 for i in range(len(keys))],labels=keys,autopct="%0.2f%%", startangle=45,radius=2)
    plt.figure(figsize=(10,4))
plt.show()
```
